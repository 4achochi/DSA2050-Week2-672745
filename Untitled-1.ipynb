```python
# Cleaning the customers dataset by removing duplicate CustomerIDs
customers_clean = customers.drop_duplicates(
    subset="CustomerID"
)

customers_clean.to_sql(
    "customers_clean",
    conn,
    if_exists="replace",
    index=False
)
```

```python
# Querying the orders and cleaned customers tables in the SQLite database to join them on CustomerID
query = """
SELECT
    o.OrderID,
    o.CustomerID,
    c.CustomerName,
    c.Segment,
    o.OrderAmount
FROM orders o
LEFT JOIN customers_clean c
    ON o.CustomerID = c.CustomerID;
"""

clean_joined = pd.read_sql_query(query, conn)
clean_joined
```

```python
# Comparing the number of rows in the original orders dataset and the cleaned joined dataset
print("Original order rows:", len(orders))
print("Clean joined rows:", len(clean_joined))

# Comparing the total sales amount in the original orders dataset and the cleaned joined dataset
print("Original sales:", orders["OrderAmount"].sum())
print("Clean joined sales:", clean_joined["OrderAmount"].sum())
```

```python
# Querying the orders and cleaned customers tables in the SQLite database to calculate key performance indicators (KPIs) by customer segment
query = """
SELECT
    COALESCE(c.Segment, 'Unmatched / NULL') AS Segment,
    COUNT(*) AS NumberOfOrders,
    SUM(o.OrderAmount) AS TotalSales,
    AVG(o.OrderAmount) AS AverageOrderValue
FROM orders o
LEFT JOIN customers_clean c
    ON o.CustomerID = c.CustomerID
GROUP BY c.Segment
ORDER BY c.Segment;
"""

segment_kpis = pd.read_sql_query(query, conn)
segment_kpis
```

**Q14. What can management learn from the segment KPIs?**

```python
## Loading the regions dataset from the JSON file
regions_df = pd.read_json("regions.json")
regions_df
```

```python
# Merging the cleaned customers dataset with the regions dataset on RegionCode
customers_regions = customers_clean.merge(
    regions_df,
    on="RegionCode",
    how="left"
)

customers_regions
```

```python
# Comparing the number of rows in the cleaned customers dataset and the customers with region names dataset
print("Clean customers:", len(customers_clean))
print("Customers with region names:", len(customers_regions))

# Displaying the customers with region names dataset with selected columns
customers_regions[
    ["CustomerID", "CustomerName", "RegionCode", "RegionName"]
]
```

**Q15. Confirm that each customer now has a readable RegionName rather than only a RegionCode.**

Yes. The JSON lookup was merged with the cleaned customer table using RegionCode, adding the corresponding RegionName for each customer.

**Q16. If a RegionCode did not exist in regions.json, what would a left merge produce for RegionName?**

The RegionName would be missing (NaN) because no matching region code would exist in the lookup table. The customer record would still be retained by the left merge.


```python
# Saving the regions dataset to the SQLite database
regions_df.to_sql(
    "regions",
    conn,
    if_exists="replace",
    index=False
)
```

```python
# Querying the orders, cleaned customers, and regions tables in the SQLite database to calculate key performance indicators (KPIs) by region
query = """
SELECT
    COALESCE(r.RegionName, 'Unmatched / Unknown') AS Region,
    COUNT(*) AS NumberOfOrders,
    SUM(o.OrderAmount) AS TotalSales
FROM orders o
LEFT JOIN customers_clean c
    ON o.CustomerID = c.CustomerID
LEFT JOIN regions r
    ON c.RegionCode = r.RegionCode
GROUP BY r.RegionName
ORDER BY r.RegionName;
"""

regional_kpis = pd.read_sql_query(query, conn)
regional_kpis
```
